```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Task 1
```

```python
days = [ "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"  ]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
plt.figure(figsize=(10,5))
plt.plot(days, temperature, marker="o", linestyle='--')
plt.title("Weekly temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.ylim(16, 28)
plt.grid(True, alpha=0.3)
for day, value in zip(days, temperature):
    plt.text(day, value + 0.3, str(value), ha="center")
plt.show()
```

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
colors = ["red", "blue", "green", "orange", "purple"]
```

```python
plt.figure(figsize=(10,5))
bars = plt.barh(languages, students, color=colors)
top = students.index(max(students))
bars[top].set_edgecolor('black')
bars[top].set_linewidth(3)
plt.gca().invert_yaxis()
plt.title('Students by Programming Language')
plt.xlabel('Number of Students')
plt.ylabel('Languages')
for i, value in enumerate(students):
    plt.text(value + 0.5, i, str(value), va='center', fontweight='bold' if i == top else 'normal')
plt.show()
```

```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_axes((0.1, 0.1, 0.7, 0.8))
sc = ax.scatter(study_hours, scores, c=ages, cmap='viridis', marker='o', s=100)
cbar = fig.colorbar(sc, ax=ax)
cbar.set_label('Age')
ax.set_title('Study Hours vs Exam Score')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Exam Score')
ax.grid(True, alpha=0.3)
best = scores.index(max(scores))
ax.scatter(study_hours[best], scores[best], s=300, facecolors='none', edgecolors='red', linewidths=2)
ax.annotate(
    'Highest score',
    xy=(study_hours[best], scores[best]),
    xytext=(study_hours[best] - 2, scores[best] - 6),
    arrowprops=dict(arrowstyle='->')
)
plt.show()
```
